## 1. 파이썬 파일 처리

```python
sample_file = 'data/sample.txt'
```

```python
# 파일을 open하여 전체 내용을 읽어서 출력

with open(sample_file, mode='rt') as f:
    content = f.read()
    print(content)

print("----- 끝 -----")
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기
with open(sample_file, 'r') as f:
    lines = f.readlines()
    print(lines)
    for line in lines:
        print(line, end="")

print("----- 끝 -----")
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며,
# 각 줄 끝의 \n도 포함해서 가져옵
with open(sample_file, 'r') as f:
    for line in f:
        print(line, end="")

print("----- 끝 -----")
```

```python
# 여러 개의 파일 읽기 (모든 txt 파일을 읽어서 출력하기)
import glob

txt_file_pattern = "data/*.txt"
files = glob.glob(txt_file_pattern)
print(files)

for file in files:
    print(file)
    with open(file, "r") as f:
        print(f.read())

print("----- 끝 -----")
```

## 4-5. 그룹별로 요약하여 데이터 확인
* value_counts()
* groupby()
* pivot_table()

```python
sample_file = 'data/sales_sample.csv'
```

```python
# 샘플 파일 로딩
import pandas as pd
sales_df = pd.read_csv(sample_file)
sales_df
```

```python
# 지역별 판매 기록의 개수
sales_df['지역'].value_counts()
```

```python
# 상품별 판매 기록의 개수와 비율
sales_df['상품'].value_counts(normalize=True)
```

```python
# 지역과 상품의 조합별 판매 기록 개수
sales_df.value_counts(['지역', '상품'])
```

```python
sample_df = pd.read_csv('data/sample_data.csv')
sample_df
```

```python
sample_df['City'].value_counts(dropna=False, ascending=True)
```

```python
# 지역별 총매출
sales_df.groupby(by = '지역')['매출'].sum()
```

```python
# 상품별 평균 매출 -> 상품을 기준으로 매출의 평균
sales_df.groupby(by = '상품')['매출'].mean()
```

```python
# 지역별 판매수량의 합계와 매출의 합계 -> 지역을 기준으로 판매수량과 각각의 매출의 합계
sales_df.groupby(by = '지역', as_index=False)[['판매수량', '매출']].sum()
```

```python
# 지역과 상품별 평균 매출 -> 지역과 상품을 기준으로 매출의 평균
sales_df.groupby(by = ['지역', '상품'], as_index=False)['매출'].mean()
```

```python
sales_df.pivot_table(
    values='매출',
    aggfunc='mean',
    index='지역',
    columns='상품'
)
```

```python

# 지역별·상품별 총매출을 구하고 행과 열의 합계 표시
sales_df.pivot_table(
    index='지역',
    columns='상품',
    values='매출',
    aggfunc='sum', 
    margins=True
)
```

```python
# 지역별 매출 : 건수, 합계, 평균, 최대값
sales_df.groupby(by='지역')['매출'].agg(['count', 'sum', 'mean', 'max'])
```
